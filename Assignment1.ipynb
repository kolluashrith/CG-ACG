### Question 2: Chromosome structures

Download the chromosome size files and make a table with the following information per species (Recall the N50 length is the length such that 50% of the genome is in chromosomes of this size or larger):

- Question 2.1. Total genome size

- Question 2.2. Number of chromosomes

- Question 2.3. Largest chromosome size and name

- Question 2.4. Smallest chromosome size and name

- Question 2.5. Mean chromosome length

- Question 2.6. N50 chromosome length

```python
def compute_n50(sizes):
    """Compute N50 given a list of chromosome/contig sizes."""
    sorted_sizes = sorted(sizes, reverse=True)
    total = sum(sorted_sizes)
    half = total / 2
    cum = 0
    for size in sorted_sizes:
        cum += size
        if cum >= half:
            return size
    return None
 
def parse_file(filepath):
    """Parse a whitespace/tab-delimited file of chrom_name<TAB>size."""
    chrom_names = []
    chrom_sizes = []
    with open(filepath) as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            parts = line.split()
            if len(parts) < 2:
                continue
            name, size = parts[0], parts[1]
            chrom_names.append(name)
            chrom_sizes.append(int(size))
    return chrom_names, chrom_sizes
 
def summarize_species(filepath):
    chrom_names, chrom_sizes = parse_file(filepath)
 
    total_size = sum(chrom_sizes)
    num_chroms = len(chrom_sizes)
 
    max_idx = chrom_sizes.index(max(chrom_sizes))
    min_idx = chrom_sizes.index(min(chrom_sizes))
 
    largest_size = chrom_sizes[max_idx]
    largest_name = chrom_names[max_idx]
 
    smallest_size = chrom_sizes[min_idx]
    smallest_name = chrom_names[min_idx]
 
    mean_length = total_size / num_chroms if num_chroms else 0
    n50 = compute_n50(chrom_sizes)
 
    # Use filename (without extension) as species identifier
    species_name = filepath.split('/')[-1].rsplit('.', 1)[0]
 
    return {
        'Species': species_name,
        'Total Genome Size': total_size,
        'Number of Chromosomes': num_chroms,
        'Largest Chromosome': largest_name,
        'Largest Chromosome Size': largest_size,
        'Smallest Chromosome': smallest_name,
        'Smallest Chromosome Size': smallest_size,
        'Mean Chromosome Length': round(mean_length, 2),
        'N50': n50
    }
 
def main():
    import glob, os, pandas as pd, matplotlib.pyplot as plt
    os.chdir("/Users/ashrithkollu/Documents/Classes/Junior_Fall/appliedgenomics2026/assignments/assignment1/")

    files = glob.glob("*.sizes")
    
    results = [summarize_species(f) for f in files]
    df = pd.DataFrame(results)
 
    print(df.to_string(index=False))
 
    # Render as a matplotlib table
    fig, ax = plt.subplots(figsize=(1.5 * len(df.columns), 0.6 * (len(df) + 1)))
    ax.axis('off')
 
    table = ax.table(
        cellText=df.values,
        colLabels=df.columns,
        cellLoc='center',
        loc='center'
    )
    table.auto_set_font_size(False)
    table.set_fontsize(9)
    table.auto_set_column_width(col=list(range(len(df.columns))))
 
    # Style header row
    for col_idx in range(len(df.columns)):
        header_cell = table[0, col_idx]
        header_cell.set_facecolor('#4C72B0')
        header_cell.set_text_props(color='white', weight='bold')
 
    plt.title("Genome Assembly Summary", fontsize=12, weight='bold', pad=20)
    plt.tight_layout()
    plt.show()
 
if __name__ == "__main__":
    main()
```

### Question 3. Random DNA shearing and fragment lengths

Before sequencing, DNA is often broken into smaller fragments. In this question, you will simulate randomly cutting a 1Mbp genome and examine the lengths of the fragments that remain. You do not need to simulate the DNA sequence itself.

- Question 3.1. In the language of your choice, simulate cutting a linear 1Mbp genome into 10,000 segments by picking 9,999 distinct cut points uniformly at random from positions 1 through 999,999. A cut at position x is between bases x and x+1. Sort the cut points, include the two ends of the genome, and compute the length of each remaining fragment. Report the mean fragment length. Plot a histogram of the fragment lengths and overlay an exponential distribution using the mean you just computed (lambda = 1/mean). How well does the exponential distribution fit the data (just a visual comparison is enough)? 

```python
#AI Disclosure: An LLM was used to generate the code for the overlay of the exponential distribution

import random as rng
import numpy as np
import matplotlib.pyplot as plt

genomesize = 1000000
num_segments = 10000

## sample distinct cut positions without replacement
cut_points = rng.sample(range(1, genomesize-1), num_segments-1)
boundaries = np.concatenate([[0], np.sort(cut_points), [genomesize]])
fragment_lengths = []

for i in range (num_segments):
  length = boundaries[i+1] - boundaries[i]
  fragment_lengths.append(length)

mean_length = np.mean(fragment_lengths)
print(f"The mean fragment length is {mean_length:.3f}")

#Sim Histogram
plt.hist(fragment_lengths, bins=100, orientation='vertical', density='True')
plt.xlabel("Fragment length (in bp)")
plt.ylabel("Fragment Probability Density")
plt.title("Histogram of DNA Shearing Fragment Lengths")

#Overlay Expo
x = np.linspace(0, np.max(fragment_lengths), 1000)
pdf = 1/mean_length * np.exp(-1/mean_length * x)
plt.plot(x, pdf, 'r-', linewidth=2, label=f'Exponential fit (mean={mean_length:.1f})')

plt.legend()
plt.show()

#The exponential distribution appears to fit the data quite well, with the histogram edges nearly lining up perfectly with the overlaid exponential distribution plot.
```

### Question 4. Coverage simulator

- Question 4.1. How many 100bp reads are needed to sequence a 1Mbp genome to 3x coverage?

Answer: You would need 3*1Mbp worth of data, which is 3000000/100 = 30000. We need 30,000 reads to achieve 3x coverage.

- Question 4.2. In the language of your choice, simulate sequencing 3x coverage of a 1Mbp genome with 100bp reads and plot the histogram of coverage. Note you do not need to actually output the sequences of the reads, you can just uniform randomly sample positions in the genome and record the coverage. You do not need to consider the strand of each read. The start position of each read should have a uniform random probability at each possible starting position (1 through 999,901). You can record the coverage in an array of 1M positions. Overlay the histogram with a Poisson distribution with lambda=3. Also overlay the distribution with a Normal distribution with a mean of 3 and a standard deviation of 1.73 (which is the square root of 3).

```python
import random as rng
import scipy.stats
import matplotlib.pyplot as plt
import numpy as np

def run_coverage_sim(genomesize, coverage, readlength):

  num_reads = int(genomesize * coverage / readlength)

  read_positions = rng.choices(range(0, genomesize-readlength+1), k=num_reads)

  ## use an array to keep track of the coverage at each position in the genome
  genome_coverage = [0] * genomesize

  for i in range(num_reads):
    
    startpos = read_positions[i]
    endpos = startpos + readlength

    for j in range (startpos, endpos):
      genome_coverage[j] = genome_coverage[j] + 1

  maxcoverage = max(genome_coverage)

  ## use an array count how many positions have 0x coverage, have 1x coverage, have 2x coverage, ...
  histogram = [0] * (maxcoverage + 1)

  for k in range(genomesize):
    cov = genome_coverage[k]
    histogram[cov] = histogram[cov] + 1

  plt.bar(list(range(maxcoverage + 1)), histogram)
  plt.xlabel("x Coverage")
  plt.ylabel("Frequency")
  plt.title(f"Coverage and Frequency with {num_reads} reads of length {readlength} bp")

  mean = coverage

  x = np.arange(0, maxcoverage + 1)
  xcont = np.linspace(0, maxcoverage, 1000)
  pmf = scipy.stats.poisson.pmf(x, mean) * genomesize
  plt.plot(x, pmf, 'r-', linewidth=2, label=f'Poisson fit (mean={mean})')
  pdf = scipy.stats.norm.pdf(xcont, loc=mean, scale = np.sqrt(mean)) * genomesize
  plt.plot(xcont, pdf, 'g-', linewidth=2, label=f'Normal fit (mean={mean})')
  plt.legend()
  plt.show()

  return histogram

histogram = run_coverage_sim(1000000, 3, 100)
```

### Question 5: Kmer Uniqueness

- Question 5.1. How many As, Cs, Gs, Ts are found in the entire chromosome? What other characters are found and how often are they found?

```python
fasta_dict = {}

file = open("/Users/ashrithkollu/Downloads/chr22.fa")

file.readline() #throw out the first line containing the name of the file

for line in file:

    line = line.strip().upper()
    for c in line:
        if not c in fasta_dict:
            fasta_dict[c] = 1
        else:
            fasta_dict[c] = fasta_dict[c] + 1

print(fasta_dict)


file.close()
```
